## Part 1 — Walkthrough

### Step 1: Import libraries

We'll use `pandas` for tabular data, `numpy` to generate synthetic values, and `matplotlib` for plotting. These are all free, open-source libraries.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# A fixed random seed means everyone generates the *same* synthetic dataset,
# which makes it easier to compare notes with classmates or ask for help.
rng = np.random.default_rng(42)
```

### Step 2: Create a small synthetic patient dataset

Real electronic health record (EHR) data is rarely handed to you as one tidy table — but almost every health dataset you will ever work with contains some version of these columns: an identifier, demographics, a few clinical measurements, and a diagnosis or outcome. This mirrors (in miniature) what you'll see in the OMOP `Person` and `Condition_Occurrence` tables next week.

```python
n = 200

df = pd.DataFrame({
    "patient_id": [f"P{i:04d}" for i in range(1, n + 1)],
    "age": rng.integers(18, 90, n),
    "sex": rng.choice(["Female", "Male"], n),
    "bmi": np.round(rng.normal(27, 5, n), 1),
    "systolic_bp": rng.integers(95, 180, n),
    "smoker": rng.choice(["Yes", "No"], n, p=[0.2, 0.8]),
    "diagnosis": rng.choice(
        ["Type 2 Diabetes", "Hypertension", "Healthy", "Asthma"],
        n, p=[0.25, 0.30, 0.30, 0.15]
    ),
})

df.head()
```

### Step 3: Inspect the data

Before computing anything, always look at the shape of the data and the type of each column. `object` columns are text/categorical; `int64`/`float64` are numeric.

```python
print(df.shape)
df.dtypes
```

### Step 4: Descriptive statistics

`.describe()` gives you count, mean, standard deviation, and quartiles for numeric columns, and counts/unique values for categorical ones when you pass `include="all"`.

```python
df.describe(include="all").round(2)
```

### Step 5: Group comparisons

A common first question in health data: *does this measurement differ across groups?* Here we compare average BMI across diagnosis categories.

```python
df.groupby("diagnosis")["bmi"].agg(["mean", "std", "count"]).round(1)
```

### Step 6: A simple visualization

A histogram of age, and a boxplot of BMI by diagnosis.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].hist(df["age"], bins=15, color="#2E6DA4", edgecolor="white")
axes[0].set_title("Age distribution")
axes[0].set_xlabel("Age")
axes[0].set_ylabel("Count")

df.boxplot(column="bmi", by="diagnosis", ax=axes[1])
axes[1].set_title("BMI by diagnosis")
axes[1].set_xlabel("Diagnosis")
axes[1].set_ylabel("BMI")
plt.suptitle("")

plt.tight_layout()
plt.show()
```
